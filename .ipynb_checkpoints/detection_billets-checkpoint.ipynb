```python
# Evaluation des 4 modèles
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
from matplotlib.gridspec import GridSpec

# Données d'exemple (à remplacer par vos résultats réels)
models = ['Régression Logistique', 'K-Means', 'KNN', 'Random Forest']
conf_matrices = [
    [[108,  2],  [1, 189]] ,  # Régression Logistique
    [[109,  1],  [2, 188]],    # K-Means
    [[105,  5],  [0, 190]],   # KNN
    [[108,  2],  [0, 190]]    # Random Forest
]

# Calcul des métriques
metrics = {
    'Accuracy': [],
    'Faux Positifs': [],
    'Faux Négatifs': []
}

for matrix in conf_matrices:
    tn, fp, fn, tp = matrix[0][0], matrix[0][1], matrix[1][0], matrix[1][1]
    metrics['Accuracy'].append((tp + tn) / (tp + tn + fp + fn))
    metrics['Faux Positifs'].append(fp)
    metrics['Faux Négatifs'].append(fn)

# Création de la figure
fig = plt.figure(figsize=(15, 8))
gs = GridSpec(2, 3, figure=fig)

# 1. Heatmaps des matrices de confusion
ax0 = fig.add_subplot(gs[:, 0])
sns.heatmap(np.mean(conf_matrices, axis=0), annot=True, fmt='.1f', cmap='Blues',
           xticklabels=['Prédit Faux', 'Prédit Vrai'],
           yticklabels=['Réel Faux', 'Réel Vrai'],
           ax=ax0)
ax0.set_title('Matrice de Confusion Moyenne')

# 2. Comparaison des erreurs
ax1 = fig.add_subplot(gs[0, 1:])
x = np.arange(len(models))
width = 0.35
bars1 = ax1.bar(x - width/2, metrics['Faux Positifs'], width, label='Faux Positifs', color='#ff7f0e')
bars2 = ax1.bar(x + width/2, metrics['Faux Négatifs'], width, label='Faux Négatifs', color='#d62728')

ax1.set_title('Analyse des Erreurs par Modèle')
ax1.set_xticks(x)
ax1.set_xticklabels(models, rotation=45)
ax1.legend()

# 3. Comparaison des performances globales
ax2 = fig.add_subplot(gs[1, 1:])
ax2.bar(models, metrics['Accuracy'], color=['#1f77b4', '#ff7f0e', '#2ca02c', '#d62728'])
ax2.set_title('Accuracy des Modèles')
ax2.set_ylim(0, 1)
for i, acc in enumerate(metrics['Accuracy']):
    ax2.text(i, acc + 0.02, f"{acc:.2%}", ha='center')

# Identification du meilleur modèle
best_idx = np.argmax(metrics['Accuracy'])
fig.text(0.5, 0.01, 
         f"★ Meilleur modèle: {models[best_idx]} (Accuracy: {metrics['Accuracy'][best_idx]:.2%}, "
         f"FP: {metrics['Faux Positifs'][best_idx]}, FN: {metrics['Faux Négatifs'][best_idx]}) ★",
         ha='center', fontsize=12, bbox={'facecolor':'gold', 'alpha':0.5})

plt.tight_layout()
plt.show()
```

Le meilleur modèle ici suggéré est le Random Forest
