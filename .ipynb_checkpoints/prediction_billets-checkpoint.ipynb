```python
 # librairies de base 
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import joblib
from sklearn.preprocessing import StandardScaler
```

```python
def predict_with_all_models(input_csv):
    # Lecture du CSV 
    data = pd.read_csv(input_csv)

    required_cols = ['diagonal', 'height_left', 'height_right', 'margin_low', 'margin_up', 'length']

    # Vérifions que les colonnes existent
    if not all(col in data.columns for col in required_cols):
        print("Colonnes manquantes")
        return None

    # Chargement du scaler
    scaler = joblib.load('scaler.sav')

    # Préparation  des données
    X_new = data[required_cols]
    X_scaled = scaler.transform(X_new)

    # Chargement  des modèles
    model_files = ['logistic_regression.sav', 'K-means.sav', 'knn-model.sav', 'random_forest.sav']

    results = data.copy()

    for file in model_files:
        model = joblib.load(file)
        pred = model.predict(X_scaled)
        model_name = file.replace('.sav', '')
        results[f'prediction_{model_name}'] = pred

    return results
```
